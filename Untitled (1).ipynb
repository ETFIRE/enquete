```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load data
df = pd.read_csv('ml_dataset_final.csv', index_col=0)
if 'LEFEVRE' in df.index:
    df = df.drop('LEFEVRE')

# Standardization
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# PCA
pca = PCA(n_components=2)
components = pca.fit_transform(df_scaled)
df['pca1'] = components[:, 0]
df['pca2'] = components[:, 1]

# Plot
plt.figure(figsize=(10, 6))
# Plot normal
normal = df[df.index.isin(['S03', 'S06', 'S07', 'S08', 'S09', 'S10'])]
plt.scatter(normal['pca1'], normal['pca2'], c='blue', label='Employés normaux', alpha=0.6)

# Plot suspects
suspects = df[df.index.isin(['S01', 'S04', 'S05'])]
plt.scatter(suspects['pca1'], suspects['pca2'], c='red', label='Suspects (S01, S04, S05)', s=100)

for i, txt in enumerate(suspects.index):
    plt.annotate(txt, (suspects.iloc[i]['pca1'], suspects.iloc[i]['pca2']), xytext=(5, 5), textcoords='offset points')

plt.title('Visualisation PCA des suspects par rapport au personnel')
plt.xlabel('Composante principale 1')
plt.ylabel('Composante principale 2')
plt.legend()
plt.grid(True)
plt.savefig('pca_suspects_plot.png')
print("Graphique PCA généré et sauvegardé dans pca_suspects_plot.png")
```

```python
import pandas as pd
from sklearn.ensemble import IsolationForest


dataset = pd.read_csv('ml_dataset_final.csv', index_col=0)


if 'LEFEVRE' in dataset.index:
    df_suspects = dataset.drop('LEFEVRE')
else:
    df_suspects = dataset


model = IsolationForest(n_estimators=100, contamination=0.3, random_state=42)
df_suspects['anomaly_score'] = model.fit_predict(df_suspects)


df_suspects['is_anomaly'] = df_suspects['anomaly_score'] == -1


print("Résultats de l'analyse ML :")
print(df_suspects[['anomaly_score', 'is_anomaly']].sort_values('anomaly_score'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chargement et préparation des données
df_trans = pd.read_csv('transactions.csv')
df_trans['date'] = pd.to_datetime(df_trans['date'])
df_trans['heure_h'] = pd.to_datetime(df_trans['heure'], format='%H:%M:%S').dt.hour

# 2. Création de la colonne Mois (format YYYY-MM)
df_trans['mois'] = df_trans['date'].dt.to_period('M').astype(str)

# 3. Définition du statut horaire (Ouvert : 9h-18h)
df_trans['statut_horaire'] = 'Ouvert'
df_trans.loc[(df_trans['heure_h'] < 9) | (df_trans['heure_h'] >= 18), 'statut_horaire'] = 'Hors Horaires'

# 4. Agrégation des données par mois et par statut
df_stats = df_trans.groupby(['mois', 'statut_horaire']).size().reset_index(name='nombre_transactions')

# 5. Visualisation
plt.figure(figsize=(14, 7))
sns.barplot(data=df_stats, x='mois', y='nombre_transactions', hue='statut_horaire', 
            palette={'Ouvert': '#3498db', 'Hors Horaires': '#e74c3c'})

# Personnalisation du graphique
plt.title("Évolution mensuelle des transactions : Heures d'ouverture vs Hors Horaires", fontsize=15)
plt.xlabel("Mois", fontsize=12)
plt.ylabel("Nombre de transactions", fontsize=12)
plt.xticks(rotation=45)
plt.legend(title="Période")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px


data = [
    ['2026-01-13', 'S02', 380000, 'Pégase'],
    ['2026-01-19', 'S02', 380000, 'Pégase'],
    ['2026-02-05', 'S02', 450000, 'Cerbère'],
    ['2026-02-20', 'S02', 450000, 'Cerbère'],
    ['2026-03-05', 'S02', 400000, 'Hermès'],
    ['2026-03-15', 'S02', 400000, 'Hermès'],
    ['2026-03-25', 'S02', 500000, 'Pégase'],
    
    ['2026-02-10', 'S07', 50000, 'Divers'],
    ['2026-03-10', 'S08', 75000, 'Divers']
]

df = pd.DataFrame(data, columns=['Date', 'Employé', 'Montant', 'Projet'])
df['Date'] = pd.to_datetime(df['Date'])


df = df.sort_values('Date')

# Calcul du cumul par employé
df['Cumul'] = df.groupby('Employé')['Montant'].cumsum()


fig = px.line(df, x='Date', y='Cumul', color='Employé',
              title='Évolution du cumul des transferts vers paradis fiscaux par employé',
              labels={'Cumul': 'Cumul Montant (EUR)', 'Date': 'Date de transaction'})


fig.update_traces(mode='lines+markers')
fig.show()
```

```python
import pandas as pd

# 1. Chargement
df_trans = pd.read_csv('transactions.csv')

# 2. Liste des paradis fiscaux de l'enquête
# KY: Caïmans, VG: Vierges Brit., PA: Panama, LU: Luxembourg, CH: Suisse
paradis_fiscaux = ['KY', 'VG', 'PA']

# 3. Filtrage des transactions suspectes
df_suspect = df_trans[df_trans['beneficiaire_pays'].isin(paradis_fiscaux)].copy()

# 4. Identification des acteurs
initiateurs = df_suspect.groupby('id_suspect_initiateur').agg({
    'montant_eur': ['count', 'sum'],
    'beneficiaire_pays': lambda x: ', '.join(x.unique())
}).reset_index()

initiateurs.columns = ['ID Employé', 'Nombre Transactions', 'Total Envoyé (€)', 'Destinations']

# 5. Affichage du classement par montant
print("--- CLASSEMENT DES EMPLOYÉS PAR MONTANT ENVOYÉ ---")
print(initiateurs.sort_values(by='Total Envoyé (€)', ascending=False).to_string(index=False))

# 6. Zoom sur le réseau de validation
print("\n--- RÉSEAU DE VALIDATION (COLLUSION) ---")
collusion = df_suspect.groupby(['id_suspect_initiateur', 'validateur_id']).size().reset_index(name='Nb_Validations')
# On affiche les duos qui ont validé le plus de transactions suspectes
print(collusion.sort_values(by='Nb_Validations', ascending=False).head(10).to_string(index=False))
```

```python
import pandas as pd

# 1. Chargement
df_emails = pd.read_csv('emails.csv')

# 2. Définition des suspects et mots-clés
mots_cles = ['Pégase', 'Cerbère', 'Hermès', 'Lefèvre', 'rapport', 'digoxine']
pattern = '|'.join(mots_cles)

# 3. Filtrage S05 + Mots-clés
# On cherche S05 en expéditeur ou destinataire ET la présence des mots-clés
mails_s05 = df_emails[(df_emails['expediteur_id'] == 'S05') | (df_emails['destinataire_id'] == 'S05')]

mails_suspects_s05 = mails_s05[
    (mails_s05['corps'].str.contains(pattern, case=False, na=False)) | 
    (mails_s05['sujet'].str.contains(pattern, case=False, na=False))
]

# 4. Affichage détaillé
print(f"--- DÉTAIL DES {len(mails_suspects_s05)} MAILS SUSPECTS DE S05 ---\n")

for index, row in mails_suspects_s05.iterrows():
    print(f"DATE: {row['timestamp']}")
    print(f"DE: {row['expediteur_id']} | À: {row['destinataire_id']}")
    print(f"SUJET: {row['sujet']}")
    print(f"MESSAGE: {row['corps']}")
    print("-" * 50)
```
